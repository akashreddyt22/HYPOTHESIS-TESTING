```python
import numpy as np
from scipy import stats


xbar = 3050          # sample mean weekly cost (Rs.)
X_mean = 600          # mean units produced per week
sigma_X = 25           # std dev of units produced

n = 25                 # sample size

# Step 1: Hypotheses
# W = 1000 + 5X  -> theoretical mean cost
mu = 1000 + 5 * X_mean
print("Theoretical mean weekly cost (mu):", mu)

print("\nStep 1: Hypotheses")
print("H0: mu = 4000  (model is still accurate)")
print("H1: mu > 4000  (weekly operating costs are higher than the model suggests)")

# Step 2: Test statistic-
sigma_W = 5 * sigma_X          # std dev of weekly cost
std_error = sigma_W / np.sqrt(n)

z = (xbar - mu) / std_error

print("\nStep 2: Test Statistic")
print("sigma_W (5*sigma_X):", sigma_W)
print("Standard Error:", std_error)
print("Z statistic:", z)

# Step 3: Critical value (alpha = 0.05, one-tailed test)

alpha = 0.05
z_critical = stats.norm.ppf(1 - alpha)

print("\nStep 3: Critical Value")
print("Alpha:", alpha)
print("Z critical (one-tailed):", z_critical)

# Step 4: Decision
print("\nStep 4: Decision")
if z > z_critical:
    decision = "Reject the null hypothesis (H0)"
else:
    decision = "Fail to reject the null hypothesis (H0)"
print(decision)

# Step 5: Conclusion
print("\nStep 5: Conclusion")
if z > z_critical:
    print("There IS strong evidence to support the restaurant owners' claim")
    print("that weekly operating costs are higher than the model suggests.")
else:
    print("There is NOT strong evidence to support the restaurant owners' claim.")
    print("The sample data does not statistically justify saying costs are higher")
    print("than what the cost model predicts.")
```
